```python
import pandas as pd
import glob
import os

# Step 1: Get all CSV file names
csv_files = glob.glob("*.csv")

# Step 2: Remove duplicate copies like (1), (2)
unique_files = []
for f in csv_files:
    if "(1)" in f or "(2)" in f or "(3)" in f:
        continue
    unique_files.append(f)

print("Files being merged:", unique_files)

# Step 3: Read & merge all files
df_list = []
for file in unique_files:
    temp = pd.read_csv(file, encoding="latin-1")   # avoids encoding error
    temp["Source_File"] = file                     # add filename column
    df_list.append(temp)

# Step 4: Combine into single dataframe
merged_df = pd.concat(df_list, ignore_index=True)

# Step 5: Show first few rows
merged_df.head()
```

```python
import pandas as pd
import numpy as np
import re
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.decomposition import TruncatedSVD
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
import umap
import matplotlib.pyplot as plt
from wordcloud import WordCloud
import nltk
nltk.download('wordnet')
nltk.download('omw-1.4')

# assume merged_df is already in the environment
print("Rows:", len(merged_df))
print("Sample NIC Name values:")
merged_df['NIC Name'].sample(10).tolist()
```

```python
# Clean & normalize NIC Name
```

```python
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()

def clean_text(s):
    if pd.isna(s): return ""
    s = str(s)
    s = s.lower()
    # common replacements / expand abbreviations
    s = s.replace('&', ' and ')
    s = re.sub(r'\bmfg\b|\bmfgs\b|\bmfgng\b', 'manufacturing', s)
    s = re.sub(r'\bdept\b', 'department', s)
    # remove punctuation/numbers but keep words
    s = re.sub(r'[^a-z\s]', ' ', s)
    s = re.sub(r'\s+', ' ', s).strip()
    # lemmatize small tokens
    tokens = [lemmatizer.lemmatize(t) for t in s.split() if len(t)>1]
    return " ".join(tokens)

merged_df['nic_clean'] = merged_df['NIC Name'].astype(str).apply(clean_text)
merged_df['nic_clean'].head(10)
```

```python
# . Rule-based keyword mapping (fast baseline)
```

```python
import re

RULES = {
    'Retail': [r'\bretail\b', r'\bshop\b', r'\bstore\b', r'\bwholesale\b', r'\bsupermarket\b', r'\boutlet\b'],
    'Manufacturing': [r'\bmanufactur', r'\bmfg\b', r'\bfabricat', r'\bfactory\b', r'\bproduction\b'],
    'Agriculture': [r'\bfarm\b', r'\bagricult', r'\bcultivat\b', r'\bhorticultur\b', r'\bagro\b'],
    'Poultry': [r'\bpoultr', r'\bchicken\b', r'\bduck\b'],
    'Construction': [r'\bconstruct', r'\bbuilder', r'\bbuilding\b', r'\bcontractor\b'],
    'Textiles': [r'\btextil', r'\bfabric\b', r'\bgarment\b', r'\bweav\b', r'\bloom\b'],
    'Chemicals': [r'\bchemical', r'\bpaint\b', r'\bfertilizer\b', r'\bpharma\b', r'\bpharmaceutical\b'],
    'Plastics & Rubber': [r'\bplastic', r'\brubber\b', r'\bpolymer\b'],
    'Food Processing': [r'\bfood\b', r'\bbeverage\b', r'\bprocessing\b', r'\bpacked\b'],
    'Furniture': [r'\bfurniture\b', r'\bwoodwork\b', r'\bwooden\b', r'\bfurnitur\b'],
    'Services': [r'\btransport\b', r'\bhotel\b', r'\brestaurant\b', r'\bservice\b', r'\blogistic\b'],
    'Mining': [r'\bmining\b', r'\bquarry\b', r'\bore\b'],
    'Other': [r'.*']  # fallback
}

def map_rulewise(text):
    for cat, patterns in RULES.items():
        for pat in patterns:
            if re.search(pat, text):
                return cat
    return 'Other'

merged_df['rule_group'] = merged_df['nic_clean'].apply(map_rulewise)
merged_df['rule_group'].value_counts().head(20)
```

```python
# TF-IDF + KMeans clustering (discover unknown groups)
```

```python
texts = merged_df['nic_clean'].fillna('').astype(str).tolist()

tfidf = TfidfVectorizer(max_features=4000, ngram_range=(1,2), stop_words='english')
X = tfidf.fit_transform(texts)

# Use TruncatedSVD for dimensionality reduction before KMeans (optional speed)
svd = TruncatedSVD(n_components=100, random_state=42)
X_red = svd.fit_transform(X)

# Choose cluster count (try 12-20). We'll use 14 as a starting point.
k = 14
km = KMeans(n_clusters=k, random_state=42, n_init=10)
clusters = km.fit_predict(X_red)
merged_df['cluster'] = clusters

# Inspect top terms per cluster
terms = tfidf.get_feature_names_out()
order = km.cluster_centers_.argsort()[:, ::-1]
for i in range(k):
    top = [terms[ind] for ind in order[i, :10]]
    print(f"Cluster {i}: ", top)
```

```python
# Visualize clusters with UMAP (2D)
```

```python
reducer = umap.UMAP(n_components=2, random_state=42)
emb = reducer.fit_transform(X_red)

plt.figure(figsize=(10,6))
scatter = plt.scatter(emb[:,0], emb[:,1], c=clusters, cmap='tab20', s=8)
plt.title("UMAP projection of industry descriptions (colored by KMeans cluster)")
plt.show()
```

```python
# Inspect cluster contents (samples) — pick clusters to label
```

```python
for i in range(k):
    print("=== Cluster", i, "sample NIC names ===")
    sample = merged_df[merged_df['cluster']==i]['NIC Name'].sample(min(8, merged_df[merged_df['cluster']==i].shape[0]), random_state=1).tolist()
    print("\n".join(sample))
    print()
```

```python
# Combine rule-based map + cluster labels
```

```python
# Example manual cluster_label mapping (you must fill after inspection)
cluster_label_map = {
    0: 'Manufacturing',
    1: 'Construction',
    2: 'Agriculture',
    3: 'Textiles',
    4: 'Retail',
    5: 'Chemicals',
    6: 'Services',
    7: 'Food Processing',
    8: 'Plastics & Rubber',
    9: 'Furniture',
    10: 'Mining',
    11: 'Poultry',
    12: 'Manufacturing',
    13: 'Other'
}

merged_df['cluster_group'] = merged_df['cluster'].map(cluster_label_map)

# Final group: prefer rule_group if not 'Other', else cluster_group, else 'Other'
def final_group(row):
    if row['rule_group'] != 'Other':
        return row['rule_group']
    if pd.notna(row.get('cluster_group')):
        return row['cluster_group']
    return 'Other'

merged_df['industry_group'] = merged_df.apply(final_group, axis=1)
merged_df['industry_group'].value_counts()
```

```python
# Train a classifier to auto-label new rows
```

```python
# Create a labeled set: use rows where rule_group != 'Other' OR cluster_group not null as proxy labels,
# but ideally hand-correct a sample and use that as gold labels.
merged_df['label_for_training'] = merged_df['industry_group']

# Remove 'Other' if you want clearer training
train_df = merged_df[merged_df['label_for_training']!='Other'].sample(frac=1, random_state=42)

y = train_df['label_for_training']
X_text = train_df['nic_clean']

vec = TfidfVectorizer(max_features=5000, ngram_range=(1,2), stop_words='english')
X_vec = vec.fit_transform(X_text)

X_train, X_test, y_train, y_test = train_test_split(X_vec, y, test_size=0.2, random_state=42, stratify=y)

clf = RandomForestClassifier(n_estimators=200, random_state=42, n_jobs=-1)
clf.fit(X_train, y_train)
pred = clf.predict(X_test)
print(classification_report(y_test, pred))
```

```python
def classify_industry(text):
    text = text.lower()

    # Retail
    if any(word in text for word in ["shop", "store", "retail", "trading"]):
        return "Retail"

    # Manufacturing
    if any(word in text for word in [
        "manufact", "factory", "production", "plastic", "rubber",
        "chemical", "cement", "steel", "fabrication"
    ]):
        return "Manufacturing"

    # Agriculture related (not cultivators)
    if any(word in text for word in ["poultry", "dairy", "milk", "farming"]):
        return "Agriculture & Allied"

    # Construction
    if any(word in text for word in ["construction", "building", "mason", "carpenter"]):
        return "Construction"

    # Transport
    if any(word in text for word in ["driver", "transport", "auto", "truck"]):
        return "Transport"

    # Services
    if any(word in text for word in ["service", "repair", "maintenance"]):
        return "Services"

    return "Other"
```

```python
merged_df["industry_group"] = merged_df["NIC Name"].apply(classify_industry)
```

```python
import streamlit as st
import pandas as pd
import plotly.express as px
import re
import json
import requests

# -------------------------------
# APP TITLE
# -------------------------------
st.title("📊 Industrial Human Resource Geo-Visualization Dashboard")
st.write("Analyze India's Industrial Workforce using NLP, EDA, and Geo-Visualization")

# -------------------------------
# FILE UPLOAD
# -------------------------------
uploaded_files = st.file_uploader(
    "Upload All State CSV Files",
    type="csv",
    accept_multiple_files=True
)

final_df = pd.DataFrame()

if uploaded_files:
    st.info("Reading & merging CSV files...")

    for file in uploaded_files:
        try:
            temp = pd.read_csv(file, encoding="utf-8")
        except UnicodeDecodeError:
            try:
                temp = pd.read_csv(file, encoding="ISO-8859-1")
            except Exception as e:
                st.error(f"Error reading {file.name}: {e}")
                continue

        # -------------------------------------------
        # EXTRACT STATE NAME FROM FILE NAME
        # -------------------------------------------
        clean_name = file.name.upper()
        match = re.search(r"STATE_([A-Z_]+)-?2011", clean_name)

        if match:
            state_name = match.group(1).replace("_", " ").title()
        else:
            state_name = "Unknown"

        temp["State"] = state_name  # <-- ADD STATE COLUMN HERE
        # -------------------------------------------

        final_df = pd.concat([final_df, temp], ignore_index=True)

    st.success("Files uploaded and merged successfully!")
    st.subheader("Preview of Merged Data")
    st.dataframe(final_df.head())

# -------------------------------
# NLP COLUMN EXTRACTION
# -------------------------------
if not final_df.empty:
    st.subheader("🔍 NLP Feature: Smart Column Search")
    user_query = st.text_input("Enter a keyword (ex: 'Main Workers', 'Urban', 'Females')")

    if user_query:
        pattern = re.compile(user_query, re.IGNORECASE)
        matched_cols = [col for col in final_df.columns if pattern.search(col)]
        st.write("### Matching Columns:")
        st.write(matched_cols if matched_cols else "No matching columns found")

# -------------------------------
# EDA SECTION
# -------------------------------
if not final_df.empty:
    st.subheader("📈 Exploratory Data Analysis")
    numeric_cols = final_df.select_dtypes(include=['int64', 'float64']).columns.tolist()

    if numeric_cols:
        column = st.selectbox("Select a numeric column to visualize", numeric_cols)
        fig = px.histogram(final_df, x=column, nbins=40, title=f"Distribution of {column}")
        st.plotly_chart(fig)
        st.write("### Summary Statistics")
        st.write(final_df[column].describe())

# -------------------------------
# GEO-VISUALIZATION SECTION
# -------------------------------
if not final_df.empty:
    st.subheader("🗺 Geo Visualization (Requires 'State' column)")
    possible_state_cols = [col for col in final_df.columns if 'state' in col.lower()]

    if possible_state_cols:
        state_col = st.selectbox("Select State Column", possible_state_cols)
        value_cols = final_df.select_dtypes(include=['int64', 'float64']).columns.tolist()
        metric = st.selectbox("Select Metric to Plot", value_cols)

        st.write("### Choropleth Map")

        # Load India states GeoJSON
        url = "https://raw.githubusercontent.com/geohacker/india/master/state/india_telengana.geojson"
        # You can replace above with a full India GeoJSON for all states
        try:
            india_geojson = requests.get(url).json()
        except Exception as e:
            st.error(f"Error loading India GeoJSON: {e}")
            india_geojson = None

        if india_geojson:
            fig = px.choropleth(
                final_df,
                geojson=india_geojson,
                locations=state_col,
                color=metric,
                featureidkey="properties.ST_NM",  # GeoJSON state property
                projection="mercator",
                title=f"{metric} by State"
            )
            fig.update_geos(fitbounds="locations", visible=False)
            st.plotly_chart(fig)
        else:
            st.warning("GeoJSON not loaded. Cannot display map.")
    else:
        st.warning("No state column found. Please verify the state column name.")

# -------------------------------
# END OF APP
# -------------------------------
```
